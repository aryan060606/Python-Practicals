```python
def zipper_file(filename, strings):
    file = open(filename, "w")

    for i in range(0, len(strings), 2):
        result = strings[i] + " " + strings[i + 1]
        file.write(result + "\n")

    file.close()



strings = ["hard", "work", "beats", "the", "talent", "if", "talent", "does", "not", "do", "hard", "work"]

zipper_file("output.txt", strings)


file = open("output.txt", "r")
print(file.read())
file.close()
```

```python
def reformat_dates(date_list):

    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in date_list:
        month, day, year = date.split("/")

        month_name = months[month]

        new_date = "Office Hours on " + month_name + " " + day + ", " + year

        result.append(new_date)

    return result



dates = ["10/30/2026", "11/06/2026", "01/09/2027"]

answer = reformat_dates(dates)

for date in answer:
    print(date)
```

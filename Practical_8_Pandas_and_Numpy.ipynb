```python
customer_data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aanya", "Kabir", "Meera", "Rohan", "Simran", "Arjun", "Isha", "Dev", "Nisha", "Karan"],
    "Age": [22, 35, 29, 42, 31, None, 27, 38, 33, 24],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male"],
    "Country": ["India", "India", "Nepal", "India", "Bhutan", "India", "India", "Nepal", None, "India"],
    "Purchase_Amount": [12500, 18000, 14500, 22000, 16000, 19500, 13200, None, 17100, 11800]
}

customers = pd.DataFrame(customer_data)
customers
```

```python
customers.head()
```

```python
customers.describe()
```

```python
customers.isnull().sum()
```

```python
customers["Age"] = customers["Age"].fillna(customers["Age"].median())
customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(customers["Purchase_Amount"].median())
customers["Country"] = customers["Country"].fillna(customers["Country"].mode()[0])

customers.isnull().sum()
```

```python
older_customers = customers[customers["Age"] > 30]
older_customers
```

```python
male_total = customers.loc[customers["Gender"] == "Male", "Purchase_Amount"].sum()
female_total = customers.loc[customers["Gender"] == "Female", "Purchase_Amount"].sum()

print("Total purchase amount for male customers:", male_total)
print("Total purchase amount for female customers:", female_total)
```

```python
customers[["Customer_ID", "Name", "Purchase_Amount"]]
```

```python
customers["Discounted_Price"] = customers["Purchase_Amount"] * 0.90
customers[["Customer_ID", "Name", "Purchase_Amount", "Discounted_Price"]]
```

```python
average_by_gender = customers.groupby("Gender")["Purchase_Amount"].mean()
average_by_gender
```

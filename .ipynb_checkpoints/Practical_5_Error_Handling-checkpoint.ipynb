```python
## Exercise 1: Handling ZeroDivisionError

# Create a Python function called safe_divide that takes two numbers,
# numerator and denominator, and performs division. Handle the
# ZeroDivisionError if the denominator is zero.
def safe_divide(numerator, denominator):
    if denominator == 0:
        print("Error: Cannot divide by Zero")
        return None
    else:
        result = numerator / denominator
        return result


numerator = float(input("Enter numerator: "))
denominator = float(input("Enter denominator: "))

result = safe_divide(numerator, denominator)
print("Result:", result)
```

```python
## Exercise 2: Handling ValueError

# Create a Python function that calculates the square root of a number.
# If a negative value is entered, handle the ValueError and display an
# appropriate error message.



def calculate_square_root(number1):
    if number1 < 0:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None
    else:
        result = number1 ** 0.5
        return result


number1 = float(input("Enter a number: "))

result = calculate_square_root(number1)
print("Square root:", result)
```

```python
## Exercise 3: Handling Generic Exceptions

# Create a Python function to calculate num / (num - 5).
# Store the result in a variable called result and handle any exception
# using the generic Exception class.


def calculate(num):
    if num == 5:
        print("An error occurred during calculation.")
    else:
        result = num / (num - 5)
        print("Result:", result)


num = float(input("Enter a number: "))

calculate(num)
```

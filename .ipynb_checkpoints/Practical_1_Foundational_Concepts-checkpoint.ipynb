```python
# Q3. Display integer, float, String, Boolean 




# Integer
integer_value = 10

# Float
float_value = 10.5

# String
string_value = "Hello World"

# Boolean
boolean_value = True

print("Integer:", integer_value)
print("Float:", float_value)
print("String:", string_value)
print("Boolean:", boolean_value)
```

```python
# Q4. Convert float to integer, String to integer, Boolean to integer 





#Float to Integer

float_value = 25.75
integer_value = int(float_value)

print("Float value:", float_value)
print("Converted integer value:", integer_value)


#String to Integer

string_value = "100"
integer_value = int(string_value)

print("String value:", string_value)
print("Converted integer value:", integer_value)


#Boolean to Integer


boolean_value = True
integer_value = int(boolean_value)

print("Boolean value:", boolean_value)
print("Converted integer value:", integer_value)
```

```python
 # Q5. Provide the simple operations performed in the calculator (addition, subtraction, multiplacation, divison)




# Taking two numbers
num1 = 20
num2 = 5

# Addition
addition = num1 + num2

# Subtraction
subtraction = num1 - num2

# Multiplication
multiplication = num1 * num2

# Division
division = num1 / num2

print("First Number:", num1)
print("Second Number:", num2)

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

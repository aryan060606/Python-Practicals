```python
# Problem 1: Buggy Code – Fix using super()
# Question

# The given code has a Person parent class and an Emp child class. The child class is not correctly calling the parent constructor. Fix the problem using super().




class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


Emp_details = Emp("Aryan", 103)

Emp_details.Display()
Emp_details.Print()
```

```python
# Problem 2: Anagram Check
# Question

# Write a Python program to check whether two strings are anagrams.

# Two words are anagrams when they contain the same letters in a different order.

str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

str1 = str1.replace(" ", "").lower()
str2 = str2.replace(" ", "").lower()

if sorted(str1) == sorted(str2):
    print("The strings are Anagrams")
else:
    print("The strings are not Anagrams")
```

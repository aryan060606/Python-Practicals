```python
# Q1. Write a Python program to check if a player Lionel Messi has more than
# 10 achievements. If the condition is true, print the player's name, sport,
# and achievements else print does not have more than 10 achievements.

player_name = "Lionel Messi"
sport = "Soccer"
achievements = 7

if achievements > 10:
    print("Player Name:", player_name)
    print("Sport:", sport)
    print("Achievements:", achievements)
else:
    print(player_name, "does not have more than 10 achievements.")
```

```python
# Q2. Write a Python program to check if a player belongs to the sport Tennis
# or has exactly 20 achievements. If the condition is true, print a success message.

players = [
    ["Serena Williams", "Tennis", 23],
    ["Lionel Messi", "Soccer", 7],
    ["Michael Phelps", "Swimming", 23],
    ["Usain Bolt", "Athletics", 8],
    ["Roger Federer", "Tennis", 20],
    ["Cristiano Ronaldo", "Soccer", 5]
]

for player in players:
    player_name = player[0]
    sport = player[1]
    achievements = player[2]

    if sport == "Tennis" or achievements == 20:
        print("Success:", player_name, "meets the criteria.")
```

```python
# Q3. Write a Python program to check if a player has less than 10 achievements
# and does not play Tennis. Print their details if they meet the criteria.

players = [
    ["Serena Williams", "Tennis", 23],
    ["Lionel Messi", "Soccer", 7],
    ["Michael Phelps", "Swimming", 23],
    ["Usain Bolt", "Athletics", 8],
    ["Roger Federer", "Tennis", 20],
    ["Cristiano Ronaldo", "Soccer", 5]
]

for player in players:
    player_name = player[0]
    sport = player[1]
    achievements = player[2]

    if achievements < 10 and sport != "Tennis":
        print("Player Name:", player_name)
        print("Sport:", sport)
        print("Achievements:", achievements)
        print()
```

```python
# Q4. Write a Python program in a Jupyter Notebook that takes an integer input
# (1 to 9999) from the user and determines the following using branching statements:
# 1. If the number is even or odd.
# 2. If the number is a palindrome.
# 3. If the number is divisible by the sum of its digits.

num = int(input("Enter an integer between 1 and 9999: "))

if 1 <= num <= 9999:

    # Check even or odd
    if num % 2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")

    # Check palindrome
    if str(num) == str(num)[::-1]:
        print("The number is a palindrome.")
    else:
        print("The number is not a palindrome.")

    # Calculate sum of digits
    digit_sum = sum(int(digit) for digit in str(num))

    # Check divisibility by sum of digits
    if num % digit_sum == 0:
        print("The number is divisible by the sum of its digits.")
    else:
        print("The number is not divisible by the sum of its digits.")

else:
    print("Invalid input. Please enter a number from 1 to 9999.")
```

```python
# Q7. Write a while loop to copy the strings of the list squares to the list
# new_squares. Stop and exit the loop if the value on the list is not 'orange'.

squares = ['red', 'yellow', 'green', 'purple', 'blue']
new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print("new_squares:", new_squares)
```

```python
# Q8. The following is a list of animals in a National Zoo.
# Your brother needs to write an essay on the animals whose names are made
# of 7 letters. Help him find those animals through a while loop and create
# a separate list of such animals.

Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])
    i += 1

print("Animals with 7 letters:", seven_letter_animals)
```

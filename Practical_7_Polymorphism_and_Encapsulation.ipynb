```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"


p1 = Point(2, 3)

print(p1)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name)


class Dog(Animal):
    pass


d1 = Dog("Rex")
d1.speak()
```

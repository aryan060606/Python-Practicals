```python
# Question 1: Create an empty Dictionary.

inventory = {}

print(inventory)
```

```python
# Question 2: Store the first product details in variables.
# Product Name = Mobile Phone
# Product Quantity = 5
# Product Price = 20000
# Product Release Year = 2020

ProductNo1_name = "Mobile Phone"
ProductNo1_quantity = 5
ProductNo1_price = 20000
ProductNo1_releaseYear = 2020

print("Product Name:", ProductNo1_name)
print("Product Quantity:", ProductNo1_quantity)
print("Product Price:", ProductNo1_price)
print("Product Release Year:", ProductNo1_releaseYear)
```

```python
# Question 3: Add details in inventory.

inventory[ProductNo1_name] = {
    "quantity": ProductNo1_quantity,
    "price": ProductNo1_price,
    "releaseYear": ProductNo1_releaseYear
}

print("Inventory:", inventory)
```

```python
# Question 4: Store the second product details in variables.
# Enter the actual Product 2 details given by your teacher.

ProductNo2_name = "Pen"
ProductNo2_quantity = 100
ProductNo2_price = 10
ProductNo2_releaseYear = 2026

print("Product Name:", ProductNo2_name)
print("Product Quantity:", ProductNo2_quantity)
print("Product Price:", ProductNo2_price)
print("Product Release Year:", ProductNo2_releaseYear)
```

```python
# Question 5: Add the second product details as an item in inventory.

inventory[ProductNo2_name] = {
    "quantity": ProductNo2_quantity,
    "price": ProductNo2_price,
    "releaseYear": ProductNo2_releaseYear
}

print("Inventory:", inventory)
```

```python
# Question 6: Display the products present in inventory.

print("Products present in inventory:")

for product in inventory:
    print(product)
```

```python
# Question 7: Check if ProductNo1_releaseYear and ProductNo2_releaseYear are in inventory.

release_years = [inventory[product]["releaseYear"] for product in inventory]

if ProductNo1_releaseYear in release_years:
    print("ProductNo1_releaseYear is present in inventory.")
else:
    print("ProductNo1_releaseYear is not present in inventory.")

if ProductNo2_releaseYear in release_years:
    print("ProductNo2_releaseYear is present in inventory.")
else:
    print("ProductNo2_releaseYear is not present in inventory.")
```

```python
# Question 8: Delete release year of both the products from the inventory.

del inventory[ProductNo1_name]["releaseYear"]
del inventory[ProductNo2_name]["releaseYear"]

print("Updated Inventory:", inventory)
```

```python
# Question 1: Create a tuple called prices containing:
# (250, 300, 150, 400, 100, 350, 200)

prices = (250, 300, 150, 400, 100, 350, 200)

print("Prices:", prices)
```

```python
# Question 2: Find and print the highest and lowest price.

highest_price = max(prices)
lowest_price = min(prices)

print("Highest price:", highest_price)
print("Lowest price:", lowest_price)
```

```python
# Question 3: Calculate and print the total sum of all prices.

total = sum(prices)

print("Total sum of all prices:", total)
```

```python
# Question 4: Convert the tuple into a sorted list in ascending order and print it.

sorted_prices = sorted(prices)

print("Sorted list:", sorted_prices)
```

```python
# Question 5: Try to modify an element in the tuple and explain what happens.

prices[0] = 500
```

```python
# Explanation:
# A tuple is immutable, which means its elements cannot be modified
# after the tuple has been created.
# Therefore, trying to modify an element of a tuple produces a TypeError.
```

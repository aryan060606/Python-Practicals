```python
# Q2. Now store the number of items in shopping list:
#    Watch, laptop, shoes, pen, clothes.

shopping_list = ["Watch", "laptop", "shoes", "pen", "clothes"]

print("Shopping List:", shopping_list)
```

```python
# Q5. Print entire shopping list.

print("Entire Shopping List:", shopping_list)
```

```python
# Q6. Print the important items like laptop and shoes from the shopping list.

print("Important Items:", shopping_list[1])
print("Important Items:", shopping_list[2])
```
